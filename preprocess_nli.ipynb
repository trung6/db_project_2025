```python
import pandas as pd

PATH = "run/accuracy/datasets/SNLI.csv"

df = pd.read_csv("tiff_results/nli_predictions_gemma3_4b.csv")
```

```python
df.head()

df.columns
```

```python
# 1. Drop
df = df.drop(columns=["raw_model_output"])

df.columns
```

```python

# 2. Rename
df = df.rename(columns={
    "predicted_label": "gemma-3-4b-it",
    "ground_truth_label": "label"
})


df.columns
```

```python

# 3. Reorder
new_order = ['original_parquet_row_index', "label", "premise", "hypothesis", "gemma-3-4b-it"]  # specify exact order
df = df[new_order]

df.columns
```

```python
df.to_csv(PATH, index=False)
```
